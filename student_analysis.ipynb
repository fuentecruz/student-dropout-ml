Gráfico simple

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x="Dropout", data=df)
plt.title("Dropout Distribution")
plt.show()
```

```python
sns.boxplot(x="Dropout", y="First_Semester_Grade", data=df)
plt.show()
```
